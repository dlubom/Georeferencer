```python
import pandas as pd
import numpy as np
import json
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid', font_scale=1.1)
plt.rcParams['figure.figsize'] = (12, 6)
plt.rcParams['figure.dpi'] = 120

# Wyświetlaj WSZYSTKO — żadnego truncowania
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
pd.set_option('display.max_colwidth', None)
pd.set_option('display.width', None)
pd.set_option('display.max_seq_items', None)
pd.set_option('display.float_format', '{:.6f}'.format)

df_caves = pd.read_parquet('parquet/CAVES.parquet')
df_users = pd.read_parquet('parquet/USERS.parquet')
df_assignments = pd.read_parquet('parquet/ASSIGNMENTS.parquet')
df_submissions = pd.read_parquet('parquet/SUBMISSIONS.parquet')

print(f'Zgłoszenia: {len(df_submissions)}, Jaskinie: {len(df_caves)}, Użytkownicy: {len(df_users)}')
```

## 2. Parsowanie punktów kliknięć użytkowników

Wyciągamy z `points_orig_json` współrzędne pikselowe:
- **punkt[0]** = otwór jaskini
- **punkt[1]** = początek paska skali
- **punkt[2]** = koniec paska skali

```python
def parse_points(json_str):
    try:
        return json.loads(json_str)
    except (json.JSONDecodeError, TypeError):
        return []

df_submissions['_points'] = df_submissions['points_orig_json'].apply(parse_points)
df_submissions['entrance_x'] = df_submissions['_points'].apply(lambda p: p[0]['x'] if len(p) > 0 else np.nan)
df_submissions['entrance_y'] = df_submissions['_points'].apply(lambda p: p[0]['y'] if len(p) > 0 else np.nan)
df_submissions['scale_start_x'] = df_submissions['_points'].apply(lambda p: p[1]['x'] if len(p) > 1 else np.nan)
df_submissions['scale_start_y'] = df_submissions['_points'].apply(lambda p: p[1]['y'] if len(p) > 1 else np.nan)
df_submissions['scale_end_x'] = df_submissions['_points'].apply(lambda p: p[2]['x'] if len(p) > 2 else np.nan)
df_submissions['scale_end_y'] = df_submissions['_points'].apply(lambda p: p[2]['y'] if len(p) > 2 else np.nan)

df_submissions['scale_bar_px'] = np.sqrt(
    (df_submissions['scale_end_x'] - df_submissions['scale_start_x'])**2 +
    (df_submissions['scale_end_y'] - df_submissions['scale_start_y'])**2
)

# Konwersja A-F na float
for col in ['A', 'B', 'C', 'D', 'E', 'F', 'pixels_per_meter', 'total_deg', 'north_deg']:
    df_submissions[col] = pd.to_numeric(df_submissions[col], errors='coerce')

print(f'Sparsowane punkty. Brak entrance_x: {df_submissions.entrance_x.isna().sum()}')
print(f'Przykład entrance (x,y): ({df_submissions.entrance_x.iloc[0]:.1f}, {df_submissions.entrance_y.iloc[0]:.1f})')
print(f'Przykład scale_bar_px: {df_submissions.scale_bar_px.iloc[0]:.1f} px')
```

## 3. Agregacja per jaskinia — odchylenia standardowe kluczowych metryk

```python
MIN_SUBS = 2
cave_counts = df_submissions.groupby('cave_id').size()
valid_caves = cave_counts[cave_counts >= MIN_SUBS].index
df = df_submissions[df_submissions.cave_id.isin(valid_caves)].copy()

metrics = ['entrance_x', 'entrance_y', 'pixels_per_meter', 'total_deg', 'C', 'F', 'scale_bar_px']

cave_stats = df.groupby('cave_id')[metrics].agg(['median', 'std', 'min', 'max', 'count'])
cave_stats.columns = ['_'.join(c) for c in cave_stats.columns]

for m in metrics:
    q1 = df.groupby('cave_id')[m].quantile(0.25)
    q3 = df.groupby('cave_id')[m].quantile(0.75)
    cave_stats[f'{m}_iqr'] = q3 - q1

print(f'Jaskinie z >= {MIN_SUBS} zgłoszeniami: {len(valid_caves)}')
print(f'Rozkład liczby zgłoszeń:')
print(cave_counts[cave_counts >= MIN_SUBS].value_counts().sort_index())
cave_stats.head()
```

## 4. Podsumowanie — rozkłady odchyleń standardowych per jaskinia

```python
summary_cols = {
    'entrance_x_std': 'Wejście X (px)',
    'entrance_y_std': 'Wejście Y (px)',
    'pixels_per_meter_std': 'Piksele/metr',
    'total_deg_std': 'Kąt obrotu (°)',
    'C_std': 'Easting C (m)',
    'F_std': 'Northing F (m)',
    'scale_bar_px_std': 'Pasek skali (px)',
}

summary = cave_stats[list(summary_cols.keys())].describe(
    percentiles=[0.25, 0.5, 0.75, 0.9, 0.95, 0.99]
).round(4)
summary.rename(columns=summary_cols, inplace=True)
display(summary)
```

## 5. Histogramy odchyleń standardowych per jaskinia (6 metryk)

```python
fig, axes = plt.subplots(2, 3, figsize=(16, 10))
titles = {
    'entrance_x_std': 'Odch. std. wejścia X (px)',
    'entrance_y_std': 'Odch. std. wejścia Y (px)',
    'pixels_per_meter_std': 'Odch. std. piksele/metr',
    'total_deg_std': 'Odch. std. kąta obrotu (°)',
    'C_std': 'Odch. std. easting C (m)',
    'F_std': 'Odch. std. northing F (m)',
}

for ax, (col, title) in zip(axes.flat, titles.items()):
    data = cave_stats[col].dropna()
    ax.hist(data, bins=50, edgecolor='black', alpha=0.7)
    ax.set_title(title)
    ax.set_ylabel('Liczba jaskiń')
    if data.max() > 0 and data.median() > 0 and data.max() / (data.median() + 1e-9) > 20:
        ax.set_xscale('log')

fig.suptitle('Rozkłady odchyleń standardowych per jaskinia', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

## 6. Spójność kliknięcia punktu otworu jaskini

Kluczowe pytanie: czy użytkownicy klikają ten sam punkt otworu? Liczymy odległość euklidesową każdego kliknięcia od mediany per jaskinia.

```python
ent_medians = df.groupby('cave_id')[['entrance_x','entrance_y']].median().rename(
    columns={'entrance_x':'ent_med_x','entrance_y':'ent_med_y'})
df = df.merge(ent_medians, on='cave_id', how='left')

df['entrance_dist_from_median'] = np.sqrt(
    (df.entrance_x - df.ent_med_x)**2 + (df.entrance_y - df.ent_med_y)**2
)

cave_ent_max = df.groupby('cave_id')['entrance_dist_from_median'].max()
cave_ent_mean = df.groupby('cave_id')['entrance_dist_from_median'].mean()

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

axes[0].hist(cave_ent_max, bins=50, edgecolor='black', alpha=0.7)
axes[0].set_xlabel('Max odl. od mediany (px)')
axes[0].set_ylabel('Liczba jaskiń')
axes[0].set_title('Rozrzut kliknięć wejścia (max odl. od mediany)')
p95 = cave_ent_max.quantile(0.95)
axes[0].axvline(p95, color='red', ls='--', label=f'95. percentyl = {p95:.1f} px')
axes[0].legend()

axes[1].scatter(cave_stats['entrance_x_std'], cave_stats['entrance_y_std'], alpha=0.4, s=15)
axes[1].set_xlabel('Odch. std. wejścia X (px)')
axes[1].set_ylabel('Odch. std. wejścia Y (px)')
axes[1].set_title('Spójność kliknięć wejścia per jaskinia')

# CDF
sorted_vals = np.sort(cave_ent_max)
cdf = np.arange(1, len(sorted_vals)+1) / len(sorted_vals)
axes[2].plot(sorted_vals, cdf)
axes[2].set_xlabel('Max odl. od mediany (px)')
axes[2].set_ylabel('Skumulowany udział jaskiń')
axes[2].set_title('CDF max. rozrzutu kliknięć wejścia')
axes[2].axhline(0.95, color='red', ls='--', alpha=0.5)

plt.tight_layout()
plt.show()

print(f'Mediana max rozrzutu wejścia: {cave_ent_max.median():.1f} px')
print(f'95. percentyl: {cave_ent_max.quantile(0.95):.1f} px')
print(f'Jaskinie z max rozrzutem > 50 px: {(cave_ent_max > 50).sum()}')
print(f'Jaskinie z max rozrzutem > 100 px: {(cave_ent_max > 100).sum()}')
```

```python
# Jaskinie z za dużym rozrzutem kliknięcia otworu (> 10 px od mediany)
ENTRANCE_THRESHOLD_PX = 10

bad_entrance_caves = cave_ent_max[cave_ent_max > ENTRANCE_THRESHOLD_PX].sort_values(ascending=False)
print(f'Jaskinie z max rozrzutem otworu > {ENTRANCE_THRESHOLD_PX} px: {len(bad_entrance_caves)}')
print()

if len(bad_entrance_caves) > 0:
    # Szczegóły: dla każdej problematycznej jaskini pokaz wszystkie kliknięcia
    rows = []
    for cave_id in bad_entrance_caves.index:
        cave_subs = df[df.cave_id == cave_id][['cave_id', 'user_id', 'submission_id',
            'entrance_x', 'entrance_y', 'entrance_dist_from_median',
            'pixels_per_meter', 'total_deg', 'C', 'F']].copy()
        cave_subs['ent_median_x'] = cave_subs['entrance_x'].median()
        cave_subs['ent_median_y'] = cave_subs['entrance_y'].median()
        rows.append(cave_subs)

    df_bad_ent = pd.concat(rows)
    df_bad_ent = df_bad_ent.sort_values(['cave_id', 'entrance_dist_from_median'], ascending=[True, False])

    # Dołącz nazwy jaskiń
    cave_names = df_caves.set_index('cave_id')['name']
    df_bad_ent = df_bad_ent.merge(cave_names, left_on='cave_id', right_index=True, how='left')

    # Przesuń 'name' na początek
    cols = ['cave_id', 'name'] + [c for c in df_bad_ent.columns if c not in ['cave_id', 'name']]
    df_bad_ent = df_bad_ent[cols]

    display(df_bad_ent)
```

## 7. Spójność skali (pixels_per_meter)

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Współczynnik zmienności skali per cave
cv_ppm = cave_stats['pixels_per_meter_std'] / cave_stats['pixels_per_meter_median'].replace(0, np.nan) * 100

axes[0].hist(cv_ppm.dropna(), bins=50, edgecolor='black', alpha=0.7)
axes[0].set_xlabel('Współczynnik zmienności skali (%)')
axes[0].set_ylabel('Liczba jaskiń')
axes[0].set_title('Rozkład CV piksele/metr per jaskinia')
p95_cv = cv_ppm.quantile(0.95)
axes[0].axvline(p95_cv, color='red', ls='--', label=f'95. percentyl = {p95_cv:.2f}%')
axes[0].legend()

# Boxplot top-20 najgorszych
worst_scale = cave_stats.nlargest(20, 'pixels_per_meter_std').index
df_worst = df[df.cave_id.isin(worst_scale)].copy()

sns.boxplot(data=df_worst, x='cave_id', y='pixels_per_meter', ax=axes[1])
axes[1].set_xticklabels(axes[1].get_xticklabels(), rotation=45, ha='right', fontsize=8)
axes[1].set_title('Top 20 jaskiń z największą rozbieżnością skali')
axes[1].set_ylabel('Piksele/metr')
axes[1].set_xlabel('')

plt.tight_layout()
plt.show()

print(f'Mediana CV skali: {cv_ppm.median():.2f}%')
print(f'Jaskinie z CV > 5%: {(cv_ppm > 5).sum()}')
print(f'Jaskinie z CV > 10%: {(cv_ppm > 10).sum()}')

print('\nTop 10 jaskiń z największą rozbieżnością piksele/metr:')
display(cave_stats.nlargest(10, 'pixels_per_meter_std')[
    ['pixels_per_meter_median','pixels_per_meter_std','pixels_per_meter_min','pixels_per_meter_max','pixels_per_meter_count']
])
```

```python
# Jaskinie z niespójnym pixels_per_meter (CV > 5%)
PPM_CV_THRESHOLD = 5

bad_ppm_caves = cv_ppm[cv_ppm > PPM_CV_THRESHOLD].sort_values(ascending=False)
print(f'Jaskinie z CV piksele/metr > {PPM_CV_THRESHOLD}%: {len(bad_ppm_caves)}')
print()

if len(bad_ppm_caves) > 0:
    rows = []
    for cave_id in bad_ppm_caves.index:
        cave_subs = df[df.cave_id == cave_id][['cave_id', 'user_id', 'submission_id',
            'pixels_per_meter', 'scale_bar_px', 'realScale_m',
            'total_deg', 'C', 'F']].copy()
        cave_subs['ppm_median'] = cave_subs['pixels_per_meter'].median()
        cave_subs['ppm_diff_pct'] = ((cave_subs['pixels_per_meter'] - cave_subs['ppm_median']) / cave_subs['ppm_median'] * 100)
        rows.append(cave_subs)

    df_bad_ppm = pd.concat(rows)
    df_bad_ppm = df_bad_ppm.sort_values(['cave_id', 'ppm_diff_pct'], ascending=[True, False])

    cave_names = df_caves.set_index('cave_id')['name']
    df_bad_ppm = df_bad_ppm.merge(cave_names, left_on='cave_id', right_index=True, how='left')
    cols = ['cave_id', 'name'] + [c for c in df_bad_ppm.columns if c not in ['cave_id', 'name']]
    df_bad_ppm = df_bad_ppm[cols]

    display(df_bad_ppm)
```

## 8. Spójność rotacji (total_deg) — wpływ strzałki północy

```python
# Konwersja skipNorth na bool
df['_skip_north'] = df['skipNorth'].astype(str).str.lower().isin(['true', '1', '1.0'])

cave_north = df.groupby('cave_id')['_skip_north'].agg(['all','any'])
cave_north.columns = ['all_skip', 'any_skip']
cave_north['group'] = np.where(cave_north.all_skip, 'Wszyscy pominęli N',
                      np.where(~cave_north.any_skip, 'Wszyscy użyli N', 'Mieszane'))

print('Grupy jaskiń wg użycia strzałki północy:')
print(cave_north.group.value_counts())

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].hist(cave_stats['total_deg_std'].dropna(), bins=50, edgecolor='black', alpha=0.7)
axes[0].set_xlabel('Odch. std. kąta obrotu (°)')
axes[0].set_ylabel('Liczba jaskiń')
axes[0].set_title('Rozkład odch. std. total_deg per jaskinia')

for grp, color in [('Wszyscy pominęli N', 'steelblue'), ('Wszyscy użyli N', 'green'), ('Mieszane', 'orange')]:
    caves_in_grp = cave_north[cave_north.group == grp].index
    data = cave_stats.loc[cave_stats.index.isin(caves_in_grp), 'total_deg_std'].dropna()
    if len(data) > 0:
        axes[1].hist(data, bins=30, alpha=0.5, label=f'{grp} (n={len(data)})', color=color)

axes[1].set_xlabel('Odch. std. kąta obrotu (°)')
axes[1].set_ylabel('Liczba jaskiń')
axes[1].set_title('Wpływ użycia strzałki północy na spójność rotacji')
axes[1].legend()

plt.tight_layout()
plt.show()
```

```python
# 8a. Jaskinie gdzie north_deg ewidentnie nie jest 0 (mediana > 1°)
# north_deg = kąt strzałki północy na planie. Powinno być ~0 (północ = góra).
# Jeśli jest daleko od 0 to albo plan jest obrócony albo user źle kliknął.

NORTH_THRESHOLD_DEG = 0.5

# Tylko submissions gdzie user NIE pominął strzałki
df_with_north = df[~df['_skip_north']].copy()

cave_north_stats = df_with_north.groupby('cave_id')['north_deg'].agg(['median', 'std', 'min', 'max', 'count'])
cave_north_stats.columns = ['north_median', 'north_std', 'north_min', 'north_max', 'north_count']

# a) Jaskinie gdzie mediana north_deg jest daleko od 0
caves_not_north = cave_north_stats[cave_north_stats.north_median.abs() > NORTH_THRESHOLD_DEG].sort_values('north_median', key=abs, ascending=False)

print(f'=== A) Jaskinie z medianą |north_deg| > {NORTH_THRESHOLD_DEG}° (plan ewidentnie nie ma północy do góry) ===')
print(f'Liczba: {len(caves_not_north)}')
print()

if len(caves_not_north) > 0:
    # Dołącz nazwy
    cave_names = df_caves.set_index('cave_id')['name']
    caves_not_north = caves_not_north.merge(cave_names, left_index=True, right_index=True, how='left')
    cols = ['name'] + [c for c in caves_not_north.columns if c != 'name']
    caves_not_north = caves_not_north[cols]
    display(caves_not_north)

    # Pokaż też szczegóły kliknięć per user dla tych jaskiń
    print('\nSzczegóły kliknięć:')
    detail_rows = []
    for cave_id in caves_not_north.index:
        cave_subs = df_with_north[df_with_north.cave_id == cave_id][
            ['cave_id', 'user_id', 'submission_id', 'north_deg', 'total_deg', 'convergence_deg']
        ].copy()
        detail_rows.append(cave_subs)

    if detail_rows:
        df_north_detail = pd.concat(detail_rows).sort_values(['cave_id', 'north_deg'])
        df_north_detail = df_north_detail.merge(cave_names, left_on='cave_id', right_index=True, how='left')
        cols = ['cave_id', 'name'] + [c for c in df_north_detail.columns if c not in ['cave_id', 'name']]
        display(df_north_detail[cols])
```

```python
# 8b. Jaskinie z niespójnym north_deg (std > 1° wśród userów którzy użyli strzałki)
NORTH_STD_THRESHOLD = 1.0

caves_inconsistent_north = cave_north_stats[cave_north_stats.north_std > NORTH_STD_THRESHOLD].sort_values('north_std', ascending=False)

print(f'=== B) Jaskinie z odch. std. north_deg > {NORTH_STD_THRESHOLD}° (niespójna strzałka północy) ===')
print(f'Liczba: {len(caves_inconsistent_north)}')
print()

if len(caves_inconsistent_north) > 0:
    cave_names = df_caves.set_index('cave_id')['name']
    caves_inconsistent_north = caves_inconsistent_north.merge(cave_names, left_index=True, right_index=True, how='left')
    cols = ['name'] + [c for c in caves_inconsistent_north.columns if c != 'name']
    caves_inconsistent_north = caves_inconsistent_north[cols]
    display(caves_inconsistent_north)

    # Szczegóły per user
    print('\nSzczegóły kliknięć:')
    detail_rows = []
    for cave_id in caves_inconsistent_north.index:
        cave_subs = df_with_north[df_with_north.cave_id == cave_id][
            ['cave_id', 'user_id', 'submission_id', 'north_deg', 'total_deg', 'convergence_deg']
        ].copy()
        detail_rows.append(cave_subs)

    if detail_rows:
        df_north_incons = pd.concat(detail_rows).sort_values(['cave_id', 'north_deg'])
        df_north_incons = df_north_incons.merge(cave_names, left_on='cave_id', right_index=True, how='left')
        cols = ['cave_id', 'name'] + [c for c in df_north_incons.columns if c not in ['cave_id', 'name']]
        display(df_north_incons[cols])

# C) Jaskinie mieszane (część skip N, część nie) TYLKO te gdzie to powoduje duży rozrzut total_deg
print('\n=== C) Jaskinie mieszane (skip/nie-skip N) z rozrzutem total_deg > 0.1° ===')
mixed_caves = cave_north[cave_north.group == 'Mieszane'].index
MIXED_TOTAL_DEG_THRESHOLD = 0.1
mixed_with_problem = cave_stats.loc[
    cave_stats.index.isin(mixed_caves) & (cave_stats['total_deg_std'] > MIXED_TOTAL_DEG_THRESHOLD)
].sort_values('total_deg_std', ascending=False)

print(f'Liczba: {len(mixed_with_problem)} (z {len(mixed_caves)} mieszanych ogółem)')
print()

if len(mixed_with_problem) > 0:
    mixed_detail = []
    for cave_id in mixed_with_problem.index:
        cave_subs = df[df.cave_id == cave_id][
            ['cave_id', 'user_id', 'submission_id', '_skip_north', 'north_deg', 'total_deg']
        ].copy()
        mixed_detail.append(cave_subs)

    df_mixed = pd.concat(mixed_detail).sort_values(['cave_id', '_skip_north', 'north_deg'])
    cave_names = df_caves.set_index('cave_id')['name']
    df_mixed = df_mixed.merge(cave_names, left_on='cave_id', right_index=True, how='left')
    cols = ['cave_id', 'name'] + [c for c in df_mixed.columns if c not in ['cave_id', 'name']]
    display(df_mixed[cols])
```

```python
# 8c. Wszystkie submissions z |north_deg| > 45° — gruby błąd lub plan obrócony
GROSS_NORTH_THRESHOLD = 45

bad_north_all = df_with_north[df_with_north.north_deg.abs() > GROSS_NORTH_THRESHOLD].copy()
cave_names = df_caves.set_index('cave_id')['name']
bad_north_all = bad_north_all.merge(cave_names, left_on='cave_id', right_index=True, how='left')

print(f'=== Submissions z |north_deg| > {GROSS_NORTH_THRESHOLD}° ===')
print(f'Liczba: {len(bad_north_all)}, jaskiń: {bad_north_all.cave_id.nunique()}, userów: {bad_north_all.user_id.nunique()}')
print()

cols = ['cave_id', 'name', 'user_id', 'submission_id', 'north_deg', 'total_deg', 'pixels_per_meter', 'C', 'F']
display(bad_north_all[cols].sort_values(['cave_id', 'north_deg']))

# Klasyfikacja: plan obrócony (wszyscy spójni) vs błąd usera (outlier)
print('\n=== Klasyfikacja per jaskinia ===')
for cave_id in sorted(bad_north_all.cave_id.unique()):
    all_subs = df[df.cave_id == cave_id]
    with_north = all_subs[~all_subs._skip_north]
    n_total = len(with_north)
    n_bad = len(with_north[with_north.north_deg.abs() > GROSS_NORTH_THRESHOLD])
    med = with_north.north_deg.median()
    std = with_north.north_deg.std()
    cname = cave_names.get(cave_id, '')
    if n_bad == n_total and std < 1:
        verdict = 'PLAN OBRÓCONY (wszyscy spójni)'
    elif n_bad < n_total:
        bad_users = with_north[with_north.north_deg.abs() > GROSS_NORTH_THRESHOLD].user_id.unique()
        verdict = f'BŁĄD USERA: {", ".join(u[:8] for u in bad_users)}'
    else:
        verdict = 'NIESPÓJNE'
    print(f'  {cave_id} ({cname}): {n_bad}/{n_total} z |N|>{GROSS_NORTH_THRESHOLD}°, mediana={med:.1f}°, std={std:.1f}° -> {verdict}')

# Userzy z powtarzającym się problemem
print('\n=== Userzy z |north_deg| > 45° — czy powtarzają błąd? ===')
for uid in sorted(bad_north_all.user_id.unique()):
    user_with_north = df_with_north[df_with_north.user_id == uid]
    user_bad = user_with_north[user_with_north.north_deg.abs() > GROSS_NORTH_THRESHOLD]
    print(f'  {uid}: {len(user_bad)}/{len(user_with_north)} ze strzałką N ma |N|>{GROSS_NORTH_THRESHOLD}°, mediana N={user_with_north.north_deg.median():.1f}°')
```

## 9. Spójność pozycji geograficznej (C, F) — najważniejsza metryka

C i F to easting/northing początku obrazu w metrach PL-1992. Łączą efekt wszystkich kliknięć (otwór + skala + rotacja). Odchylenie w metrach = bezpośredni błąd pozycyjny.

```python
df['geo_dist_from_median'] = np.sqrt(
    (df.C - df.groupby('cave_id')['C'].transform('median'))**2 +
    (df.F - df.groupby('cave_id')['F'].transform('median'))**2
)

cave_geo_max = df.groupby('cave_id')['geo_dist_from_median'].max()

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

axes[0].hist(cave_geo_max, bins=50, edgecolor='black', alpha=0.7)
axes[0].set_xlabel('Max odchylenie od mediany (m)')
axes[0].set_ylabel('Liczba jaskiń')
axes[0].set_title('Rozrzut geogr. początku per jaskinia (max)')
p95_geo = cave_geo_max.quantile(0.95)
axes[0].axvline(p95_geo, color='red', ls='--', label=f'95p = {p95_geo:.2f} m')
axes[0].legend()

axes[1].scatter(cave_stats['C_std'], cave_stats['F_std'], alpha=0.4, s=15)
axes[1].set_xlabel('Odch. std. C / easting (m)')
axes[1].set_ylabel('Odch. std. F / northing (m)')
axes[1].set_title('Spójność C vs F per jaskinia')

# CDF
sorted_vals = np.sort(cave_geo_max.dropna())
cdf = np.arange(1, len(sorted_vals)+1) / len(sorted_vals)
axes[2].plot(sorted_vals, cdf)
axes[2].set_xlabel('Max odchylenie od mediany (m)')
axes[2].set_ylabel('Skumulowany udział jaskiń')
axes[2].set_title('CDF max. odchylenia geograficznego')
axes[2].axhline(0.95, color='red', ls='--', alpha=0.5)
axes[2].set_xscale('log')

plt.tight_layout()
plt.show()

print(f'Mediana max odchylenia geogr.: {cave_geo_max.median():.3f} m')
print(f'95. percentyl: {cave_geo_max.quantile(0.95):.3f} m')
print(f'Jaskinie z max odchyleniem > 1 m: {(cave_geo_max > 1).sum()}')
print(f'Jaskinie z max odchyleniem > 10 m: {(cave_geo_max > 10).sum()}')
print(f'Jaskinie z max odchyleniem > 100 m: {(cave_geo_max > 100).sum()}')
```

## 10. Detekcja outlierów (IQR per jaskinia)

Flagujemy submissions gdzie dowolna kluczowa metryka wypada poza 1.5*IQR od Q1/Q3 danej jaskini.

```python
outlier_metrics = ['pixels_per_meter', 'total_deg', 'C', 'F', 'entrance_x', 'entrance_y']

df['is_outlier'] = False
df['outlier_reasons'] = ''

for m in outlier_metrics:
    q1 = df.groupby('cave_id')[m].transform(lambda x: x.quantile(0.25))
    q3 = df.groupby('cave_id')[m].transform(lambda x: x.quantile(0.75))
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    is_out = (df[m] < lower) | (df[m] > upper)
    df['is_outlier'] = df['is_outlier'] | is_out
    df.loc[is_out, 'outlier_reasons'] += m + '; '

n_outliers = df.is_outlier.sum()
n_total = len(df)
print(f'Outlierów: {n_outliers} / {n_total} ({100*n_outliers/n_total:.1f}%)')
print(f'Jaskinie z >= 1 outlierem: {df[df.is_outlier].cave_id.nunique()} / {df.cave_id.nunique()}')

print('\nKtóre metryki najczęściej powodują flagę:')
for m in outlier_metrics:
    cnt = df.outlier_reasons.str.contains(m).sum()
    print(f'  {m}: {cnt} ({100*cnt/n_total:.1f}%)')
```

## 12. Jakość per użytkownik — ranking outlierów

```python
user_stats = df.groupby('user_id').agg(
    n_submissions=('submission_id', 'count'),
    n_outliers=('is_outlier', 'sum'),
).assign(outlier_rate=lambda x: x.n_outliers / x.n_submissions * 100)

user_stats = user_stats.sort_values('outlier_rate', ascending=False)

fig, ax = plt.subplots(figsize=(14, 8))
top_users = user_stats[user_stats.n_submissions >= 5].head(20)
bars = ax.barh(range(len(top_users)), top_users.outlier_rate, color='coral', edgecolor='black')
ax.set_yticks(range(len(top_users)))
ax.set_yticklabels([f"{uid[:8]}... (n={row.n_submissions})"
                     for uid, row in top_users.iterrows()])
ax.set_xlabel('Odsetek outlierów (%)')
ax.set_title('Użytkownicy z najwyższym odsetkiem outlierów (min. 5 zgłoszeń)')
ax.invert_yaxis()

for i, (_, row) in enumerate(top_users.iterrows()):
    ax.text(row.outlier_rate + 0.3, i, f'{row.outlier_rate:.1f}% ({int(row.n_outliers)}/{int(row.n_submissions)})',
            va='center', fontsize=9)

plt.tight_layout()
plt.show()

print('\nTop 15 użytkowników:')
display(user_stats.head(15))
```

## 13. Profil odchyleń per user — heatmap (modified z-score)

```python
for m in outlier_metrics:
    cave_med = df.groupby('cave_id')[m].transform('median')
    cave_mad = df.groupby('cave_id')[m].transform(lambda x: (x - x.median()).abs().median())
    df[f'{m}_mzscore'] = np.where(cave_mad > 0, 0.6745 * (df[m] - cave_med) / cave_mad, 0)

user_deviations = df.groupby('user_id')[[f'{m}_mzscore' for m in outlier_metrics]].apply(
    lambda x: x.abs().mean()
)
user_deviations.columns = [c.replace('_mzscore','') for c in user_deviations.columns]

# Filtr: aktywni użytkownicy (>= 10 zgłoszeń)
active_users = user_stats[user_stats.n_submissions >= 10].index
ud = user_deviations.loc[user_deviations.index.isin(active_users)].copy()
ud['user_label'] = [uid[:8] + '...' for uid in ud.index]
ud = ud.set_index('user_label')

fig, ax = plt.subplots(figsize=(10, max(6, len(ud)*0.45)))
sns.heatmap(ud[outlier_metrics].sort_values('pixels_per_meter', ascending=False),
            cmap='YlOrRd', annot=True, fmt='.1f', ax=ax, linewidths=0.5)
ax.set_title('Średni |modified z-score| per metryka per użytkownik (aktywni >= 10 zgłoszeń)')
ax.set_ylabel('Użytkownik')
plt.tight_layout()
plt.show()
```

## 16. Composite quality score per jaskinia

Łączymy odchylenia standardowe w jeden wskaźnik jakości (0 = doskonała zgodność, 1 = brak zgodności).

```python
quality_metrics = ['entrance_x_std', 'entrance_y_std', 'pixels_per_meter_std',
                   'total_deg_std', 'C_std', 'F_std']

for m in quality_metrics:
    cave_stats[f'{m}_pctrank'] = cave_stats[m].rank(pct=True)

cave_stats['quality_score'] = cave_stats[[f'{m}_pctrank' for m in quality_metrics]].mean(axis=1)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].hist(cave_stats.quality_score, bins=50, edgecolor='black', alpha=0.7)
axes[0].set_xlabel('Wskaźnik jakości (0=doskonała, 1=najgorsza)')
axes[0].set_ylabel('Liczba jaskiń')
axes[0].set_title('Rozkład wskaźnika jakości georeferencji')
axes[0].axvline(0.9, color='red', ls='--', label='Próg problematycznych (0.9)')
axes[0].legend()

# Top 20 worst caves
worst20 = cave_stats.nlargest(20, 'quality_score')
axes[1].barh(range(len(worst20)), worst20.quality_score, color='coral', edgecolor='black')
axes[1].set_yticks(range(len(worst20)))
axes[1].set_yticklabels(worst20.index)
axes[1].set_xlabel('Quality score')
axes[1].set_title('Top 20 najgorszych jaskiń')
axes[1].invert_yaxis()

plt.tight_layout()
plt.show()

problem_count = (cave_stats.quality_score > 0.9).sum()
print(f'Jaskinie z quality_score > 0.9: {problem_count}')
print(f'Jaskinie z quality_score < 0.5 (dobre): {(cave_stats.quality_score < 0.5).sum()}')

print('\nTop 10 najgorszych:')
display(cave_stats.nlargest(10, 'quality_score')[quality_metrics + ['quality_score']])
```
